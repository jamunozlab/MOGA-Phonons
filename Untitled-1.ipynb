```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import yaml
from math import ceil

DATA_DIR = Path("/Users/jamunoz/Documents/GitHub/MOGA-Phonons/dataframes")

df13 = pd.read_pickle(DATA_DIR / "dataframe000013.pkl")
df14 = pd.read_pickle(DATA_DIR / "dataframe000014.pkl")

df13["run_id"] = "run_013"
df14["run_id"] = "run_014"

df = pd.concat([df13, df14], ignore_index=True)

df.columns
```

```python
# Rename these if your dataframe uses slightly different column names
mass_col = "mass"
alat_col = "a_val"
fitness_col = "fitness_norm"

# If needed, uncomment and inspect:
# df[[mass_col, alat_col, fitness_col]].head()
```

```python
top5 = (
    df.sort_values(fitness_col, ascending=False)
      .groupby(["run_id", mass_col, alat_col], as_index=False)
      .head(5)
      .reset_index(drop=True)
)

top5.groupby(["run_id", mass_col, alat_col]).size().head()
```

```python
def get_yaml_path(row):
    """
    Adjust this function if your dataframe stores paths differently.
    """
    possible_path_cols = [
        "yaml_path",
        "band_yaml_path",
        "band_ga_yaml",
        "filepath",
        "file_path",
        "path",
    ]

    for col in possible_path_cols:
        if col in row.index and pd.notna(row[col]):
            return Path(row[col])

    if "simulation_path" in row.index and "generation" in row.index:
        gen = int(row["generation"])
        return Path(row["simulation_path"]) / "phonon_generations" / f"generation_{gen:06d}" / "band_ga.yaml"

    if "sim_path" in row.index and "generation" in row.index:
        gen = int(row["generation"])
        return Path(row["sim_path"]) / "phonon_generations" / f"generation_{gen:06d}" / "band_ga.yaml"

    raise ValueError("Could not infer band_ga.yaml path from dataframe row.")
```

```python
def load_phonopy_band_yaml(yaml_path):
    """
    Returns x positions, frequencies, high-symmetry labels, and label positions.
    """
    yaml_path = Path(yaml_path)

    with open(yaml_path, "r") as f:
        data = yaml.safe_load(f)

    phonons = data["phonon"]

    x = np.array([p.get("distance", i) for i, p in enumerate(phonons)])

    freqs = []
    for p in phonons:
        freqs.append([band["frequency"] for band in p["band"]])

    freqs = np.array(freqs)

    labels = []
    label_positions = []

    for i, p in enumerate(phonons):
        if "label" in p and p["label"] is not None:
            labels.append(p["label"])
            label_positions.append(x[i])

    return x, freqs, labels, label_positions
```

```python
groups = sorted(top5.groupby([mass_col, alat_col]).groups.keys())

ncols = 3
nrows = ceil(len(groups) / ncols)

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(4.2 * ncols, 3.2 * nrows),
    sharey=True
)

axes = np.array(axes).reshape(-1)

for ax, (mass, alat) in zip(axes, groups):

    sub = top5[
        (top5[mass_col] == mass) &
        (top5[alat_col] == alat)
    ]

    for _, row in sub.iterrows():
        yaml_path = get_yaml_path(row)

        try:
            x, freqs, labels, label_positions = load_phonopy_band_yaml(yaml_path)
        except Exception as e:
            print(f"Skipping {yaml_path}: {e}")
            continue

        for branch in range(freqs.shape[1]):
            ax.plot(x, freqs[:, branch], color="red", alpha=0.35, linewidth=0.9)

    ax.axhline(0, color="black", linewidth=0.6)
    ax.set_title(f"a={alat}, m={mass}", fontsize=8)

    if labels and label_positions:
        ax.set_xticks(label_positions)
        ax.set_xticklabels(labels, fontsize=8)

    ax.set_ylabel("Frequency (THz)")
    ax.grid(True, alpha=0.3)

for ax in axes[len(groups):]:
    ax.axis("off")

fig.suptitle("Top 5 solutions from each run for each mass and lattice parameter", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
def get_bcc_tick_positions(x):
    """
    For a path Gamma-H-P-Gamma-N with four equal q-path segments
    sampled uniformly.
    """
    x = np.asarray(x)

    tick_positions = [
        x[0],
        x[len(x)//4],
        x[len(x)//2],
        x[3*len(x)//4],
        x[-1],
    ]

    tick_labels = [r"$\Gamma$", "H", "P", r"$\Gamma$", "N"]

    return tick_positions, tick_labels
```

```python
def fmt_sig(x, sig=3):
    return f"{float(x):.{sig}g}"
```

```python
run_colors = {
    "run_013": "tab:blue",
    "run_014": "tab:orange",
}

hs_labels = [r"$\Gamma$", "H", "P", r"$\Gamma$", "N"]

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(4.2 * ncols, 3.2 * nrows),
    sharey=False
)

axes = np.array(axes).reshape(-1)

for ax, (mass, alat) in zip(axes, groups):

    sub = top5[
        (top5[mass_col] == mass) &
        (top5[alat_col] == alat)
    ]

    for _, row in sub.iterrows():
        yaml_path = get_yaml_path(row)

        try:
            x, freqs, labels, label_positions = load_phonopy_band_yaml(yaml_path)
        except Exception as e:
            print(f"Skipping {yaml_path}: {e}")
            continue

        color = run_colors.get(row["run_id"], "gray")

        for branch in range(freqs.shape[1]):
            ax.plot(
                x,
                freqs[:, branch],
                color=color,
                alpha=0.35,
                linewidth=0.9
            )

    ax.axhline(0, color="black", linewidth=0.6)

    ax.set_title(
        f"a = {fmt_sig(alat)} Å, m = {fmt_sig(mass)} amu",
        fontsize=8
    )

    tick_positions, tick_labels = get_bcc_tick_positions(x)

    tick_positions, tick_labels = get_bcc_tick_positions(x)

    ax.set_xticks(tick_positions)
    ax.set_xticklabels(tick_labels, fontsize=8)

    ax.axhline(0, color="black", linewidth=0.6)

    ax.set_title(
        f"a = {fmt_sig(alat)} Å, m = {fmt_sig(mass)} amu",
        fontsize=8
    )

    ax.set_ylabel("Frequency (THz)")
    ax.tick_params(axis="y", labelleft=True)

    # Fixed vertical axis
    ax.set_ylim(0, 10)
    ax.set_yticks(np.arange(0, 11, 2))

    ax.grid(True, alpha=0.3)

# One clean shared legend
handles = [
    plt.Line2D([0], [0], color=run_colors["run_013"], lw=2, label="dataframe000013"),
    plt.Line2D([0], [0], color=run_colors["run_014"], lw=2, label="dataframe000014"),
]

fig.legend(
    handles=handles,
    loc="upper center",
    ncol=2,
    frameon=False,
    bbox_to_anchor=(0.5, 1.005)
)

fig.suptitle(
    "Top 5 solutions from each run for each mass and lattice parameter",
    fontsize=14,
    y=1.02
)

plt.tight_layout()
plt.show()
```
